Z funkcji `open` możemy korzystać zarówno jako klasycznej funkcji, jak i menedżera kontekstu. Ta druga metoda ceduja na Pythona zadanie zamknięcia raz otwartego pliku.

## 2*. 
#### Każdy znak w tekście ma swój indeks (nawet spacja), poniżej sprawdzamy, od jakiego indeksu zaczyna się nasz właściwy tekst.


Możemy wykorzystać elementy tekstu, aby odnaleźć początek faktycznej treści utworu:

```python
first_line = "/ Ściana tylna sceny przedstawia zamek Atrydów w Argos. /"
text_start = txt.index(first_line)
print(text_start)
```

### [-2000: ] bierzemy od ostatnich 2000 znaków do końca

```python
print(txt[-2000:])
```

### Liczymy ile razy w "txt" pojawiła się dana linijka

```python
last_line = "Wy nućcie pean błogi!"
txt.count(last_line)
```

Chcemy postąpić analogicznie w wypadku końca tekstu, jednak ostatnia fraza pojawia się w nim dwa razy. Musimy odnaleźc indeks ostatniego wystąpienia:

```python
first_occ = txt.index(last_line)  # indeks pierwszego powtórzenia naszego zdania

line_len = len(last_line)         # długość tego ile nasza fraza ma znaków

after_first = txt[first_occ + line_len:] # jesteśmy na pierwszej literze naszego zdania i dodajemy do tego długość naszego zdania by je przeskoczyć
                                         # after_first to tekst po naszym pierwszym wystapieniu az do końca

second_occ = first_occ + line_len + after_first.index(last_line) # dodajemy indeks + dlugość naszego zdania + indeks w nowym tekście
                                                   # aby zobaczyć prawdziwy indeks naszego drugiego powtórzenia (w oryginalnym tekście txt)

text_end = second_occ+line_len # wiemy na jakim indeksie zaczyna się nasz tekst, dodajemy do tego długość tekstu i wychodzi nam indeks ostatniego znaku tekstu

print(txt[text_end - 1000:text_end]) # printujemy tekst od ostatnich 1000 znaków do ostatniego znaku (indeksu)

```

Tekst ma zgrubsza uporządkowaną strukturę:

1. Wypowiedzi przedzielone są trzema znakami nowej linii: `\n`
1. Każda wypowiedź jest poprzedzona nazwą mówcy, po której następują dwa znaki nowej linii
1. Strofy wypowiedzi są przedzielone podwójnymi znakami nowej linii

Wpierw odetnijmy nagłówki i stopkę, i usuńmy okalające białe znaki metodą `strip`:

### Metoda .strip() usuwa wszystkie białe znaki (spacje, tabulatory \t oraz znaki nowej linii \n) znajdujące się wyłącznie na samym początku oraz na samym końcu tekstu.
#### [pusta linijka]   <-- ZOSTANIE USUNIĘTA (jest na samym początku)
#### Pierwszy wers
#### [pusta linijka]   <-- ZOSTAJE (jest w środku, między wersami)
#### Drugi wers
#### [pusta linijka]   <-- ZOSTANIE USUNIĘTA (jest na samym końcu)

```python
core_text = txt[text_start:text_end].strip()
```

Następnie wykorzystajmy zauważone prawidłowości do podziału tekstu na mowy.

W wypadku gdy nasze spostrzeżenia nie zostaną spełnione przez fragment tekstu, pomijamy go:

```python
data = [] # pusta lista do której będziemy wrzucać gotowe, przetworzone wypowiedzi

speeches = core_text.split("\n\n\n") # Tnie cały tekst na pojedyncze wypowiedzi w miejscach, gdzie występują trzy znaki nowej linii (\n\n\n). 
                                     # Tworzy się lista speeches, w której każdy element to jedna wypowiedź razem z nazwą postaci

for speech in speeches:
    if speech:                       # Jeśli dany speech nie jest pustym tekstem to 
        try:
            speaker, content = speech.split("\n\n", 1) # Tniemy nasze speeche na dwie części, robimy jedno cięcie w każdym speechu
                                                       # dzięki temu osoba mówiaca trfia do "speaker" a to co mówi do "content"
        except ValueError:
            # Jeśli tekst będzie nietypowy to skipujemy go 
            continue # i kontynuujemy
            
        chunks = content.split("\n\n")                # dzieli wypowiedź na pojedyncze zdania, tam gdzie jest \n\n
        data.append((speaker, chunks))                # dodajemy do listy krotkę, w której mamy speakera i 

print(f"{len(data)}/{len(speeches)}") # ile speechy udało nam się poprawnie przetworzyć
print(data[100])
```

```python
#4. podziel tekst zgodnie z zaobserwowaną strukturą pliku
#5. skonstruuj listę par (mówca, strofy)

data_a = [] 
speeches_a = core_text_a.split("\n\n\n")

for speech in speeches_a:
    if speech:
        try:
            speaker_a, content_a = speech.split("\n\n", 1)
        except ValueError:
            continue
            
    chunks = content_a.split("\n\n")
    data_a.append((speaker_a, chunks))
        
```

## Ćwiczenie 2:

Przygotuj dwa rankingi:
1. Mówców o największej LICZBIE mów
2. Mówców o najwyższej SUMIE DŁUGOŚCI mów (liczonej w znakach)

```python
#TODO
# ranking powinien mieć postać wyprintowanego stringa
# gdzie w każdej linii znajduje się numer pozycji, nazwa mówcy, oraz kwalifikujący go wynik

# 1.
speeches_num = {}
for speaker_a, chunks in data_a:
    if speaker_a not in speeches_num:
        speeches_num[speaker_a] = 0
    speeches_num[speaker_a] += 1

pairs = []

for speaker_a, num in speeches_num.items(): # .items wyciaga nam ze słownika (etykieta, wartość)
    pairs.append((num, speaker_a))  # .append() tylko dla list []

p_sorted = sorted(pairs, reverse = True)

ranking_1 = ""
position = 1

for score, speaker_a in p_sorted:
    ranking_1 += f"{position}. {speaker_a}: {score}\n"
    position += 1

print(ranking_1)
```

```python
speeches_len = {}
for speaker_a, chunks in data_a:
    if speaker_a not in speeches_len:
        speeches_len[speaker_a] = 0
    for strofa in chunks:
        speeches_len[speaker_a] = len(strofa)

pairs_2 = []

for speaker, num in speeches_len.items():
    pairs_2.append((num, speaker))

p2_sorted = sorted(pairs_2, reverse = True)

ranking_2 = ""
position = 1

for num, speaker_a in p2_sorted:
    ranking_2 += f" {position}. {speaker_a}: {num}\n"
    position += 1

print(ranking_2)
```
